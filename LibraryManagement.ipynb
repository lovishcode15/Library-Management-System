```python
#Object Oriented programming projects
#Library Management
class books:
    def __init__(self, title, author):
        self.title=title
        self.author=author
        self.available=True

    def Availability(self):
        if self.available:
            return f"{self.title} by {self.author} is available"
        else:
            return "borrowed"

class Member:
    def __init__(self, name, member_id):
        self.name=name
        self.member_id=member_id
        self.borrowed_books=[]

    def borrow_books(self, book):
        if book.available == True:
            book.available = False
            self.borrowed_books.append(book)
            return f"{self.name} has borrowed {book.title} by {book.author}"
        else:
            return f"{book.title} is not available"

    def return_books(self, book):
        if book in self.borrowed_books:
            book.available=True
            self.borrowed_books.remove(book)
            return f"{self.name} has returned {book.title} by {book.author}"
        else:
            return f"{self.name} has not borrowed {book.title} by {book.author}"

class library:
    def __init__(self):
        self.books=[]
        self.members=[]

    def add(self, book):
        self.books.append(book)
        print(f"{book.title} added in the library")

    def add_member(self, member):
        self.members.append(member)
        print(f"{member.name} added in library")

    def find_book(self, title):
        for book in self.books:
            if book.title.lower()==title.lower():
                return book
        return None

    def find_member(self, member_id):
        for member in self.members:
            if member_id == member.member_id:
                return member
        return None

library_inst = library()

book1 = books("The Alchemist", "Paulo Coelho")
book2 = books("To Kill a Mockingbird", "Harper Lee")
book3 = books("1984", "George Orwell")

library_inst.add(book1)
library_inst.add(book2)
library_inst.add(book3)

member1 = Member("Alice", "M001")
member2 = Member("Bob", "M002")

library_inst.add_member(member1)
library_inst.add_member(member2)

while True:
    print("\n---Library Management System----")
    print("1. borrow")
    print("2. return")
    print("3. exit")


    choice=input("enter how we can assist you:")

    if choice=="1":
        member_id=input("enter member id:")
        title=input("enter book title:")


        member=library_inst.find_member(member_id)
        book=library_inst.find_book(title)

        if member is None:
            print("member not found")
        elif book is None:
            print("book not found")
        else:
            print(member.borrow_books(book))

    if choice=="2":
        member_id=input("enter member id:")
        title=input("enter book title:")

        member=library_inst.find_member(member_id)
        book=library_inst.find_book(title)

        if member is None:
            print("member not found")

        elif book in None:
            print("book not found")
        else:
            print(member.return_books(book))

    if choice=="exit":
        break
```
